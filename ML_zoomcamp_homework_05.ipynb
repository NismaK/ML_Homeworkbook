### Q3

```python
import pickle
import pandas as pd

lead = {
    "lead_source": "paid_ads",
    "industry": "technology",
    "employment_status": "employed",
    "location": "north_america",
    "number_of_courses_viewed": 2,
    "annual_income": 79276.0,
    "interaction_count": 4,
    "lead_score": 0.41
}

with open("pipeline.bin", "rb") as f:
    model = pickle.load(f)

# DictVectorizer expects a list of dictionaries, not a pandas DataFrame
p = model.predict_proba([lead])[0, 1]
print(f"Q3: {p:.3f}")
```

### Q4

```python
import json
import subprocess
import time
import urllib.request

# Start the API in the background
api = subprocess.Popen(
    ["python", "-m", "uvicorn", "predict:app", "--host", "0.0.0.0", "--port", "9696"],
    stdout=subprocess.DEVNULL,
    stderr=subprocess.DEVNULL
)
time.sleep(3)

lead = {
    "lead_source": "organic_search",
    "industry": "technology",
    "employment_status": "employed",
    "location": "europe",
    "number_of_courses_viewed": 4,
    "annual_income": 80304.0,
    "interaction_count": 7,
    "lead_score": 0.74
}

req = urllib.request.Request(
    "http://localhost:9696/predict",
    data=json.dumps(lead).encode(),
    headers={"Content-Type": "application/json"},
    method="POST"
)

with urllib.request.urlopen(req) as r:
    result = json.load(r)

print(f"Q4: {result['conversion_probability']:.3f}")
print("conversion:", result["conversion"])
```
