```python
import pandas as pd
import numpy as np
# Q2
pd.__version__
```

```python
df=pd.read_csv("/content/car_fuel_efficiency_2026.csv")
```

```python
df.head()
```

```python
#Q2. Records count
print("Q2. Records count:", len(df))
```

```python
# Q3. Unique fuel types (if available)
# Dataset might have 'origin' or 'fuel' column
if 'fuel_type' in df.columns:
    print("Q3. Fuel types:", df['fuel_type'].unique())
else:
    print("Q3. Fuel types: Column not found, maybe check dataset")
```

```python
# Q4. Missing values
print("Q4. Missing values (per column):")
print(df.isnull().sum())
```

```python
print("Q5. Max fuel efficiency (fuel_efficiency_mpg):", df['fuel_efficiency_mpg'].max())
```

```python
# Q6. Median horsepower trend
# Compare median horsepower before 1980 and after 1980
if 'horsepower' in df.columns:
    median_before = df[df['model_year'] < 1980]['horsepower'].median()
    median_after = df[df['model_year'] >= 1980]['horsepower'].median()
    print("Median horsepower before 1980:", median_before)
    print("Median horsepower after 1980:", median_after)
    if median_after > median_before:
        print("Q6:", "Yes, it increased")
    elif median_after < median_before:
        print("Q6:", "Yes, it decreased")
    else:
        print("Q6:", "No")
else:
    print("Horsepower column not found")
```

```python
if 'vehicle_weight' in df.columns:
    # Normalize by dividing by the maximum value and then sum
    normalized_vehicle_weight_sum = (df['vehicle_weight'] / df['vehicle_weight'].max()).sum()
    print("Q7. Sum of normalized vehicle weight:", round(normalized_vehicle_weight_sum, 3))
else:
    print("vehicle_weight column not found")
```

```python
missing_vehicle_weight = df['vehicle_weight'].isnull().sum()
print(f"Missing values in 'vehicle_weight': {missing_vehicle_weight}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.histplot(df['vehicle_weight'], kde=True, bins=30)
plt.title('Distribution of Vehicle Weight')
plt.xlabel('Vehicle Weight')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)
plt.show()
```
